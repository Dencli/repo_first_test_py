```python
s = input()
if '_' in s:
    snake_case = s.split("_")
    for i in range(1, len(snake_case)):
        snake_case[i] = snake_case[i].title()
    print(''.join(snake_case))
elif '-' in s:
    kebabcase  = s.split("-")
    for j in range(1, len(kebabcase)):
        kebabcase[j] = kebabcase[j].title()
    print(''.join(kebabcase))
else:
    print(s)
```
